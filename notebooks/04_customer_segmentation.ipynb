# 04 Customer Segmentation

This notebook turns churn probabilities into risk segments, revenue at risk, business segments, and retention actions.

```python
import sys
from pathlib import Path
import joblib
import pandas as pd

PROJECT_ROOT = Path.cwd().parent if Path.cwd().name == 'notebooks' else Path.cwd()
sys.path.append(str(PROJECT_ROOT))

from src.config import BUSINESS_SUMMARY_PATH, CUSTOMER_FEATURES_PATH, CUSTOMER_PREDICTIONS_PATH, CUSTOMER_SEGMENTS_PATH, FEATURE_IMPORTANCE_PATH, MODEL_PATH
from src.reporting import create_business_figures
from src.segmentation import build_customer_segments, create_business_summary, create_customer_predictions
```

## Generate Customer Probabilities

The model probability estimates the chance that each customer will churn.

```python
features = pd.read_csv(CUSTOMER_FEATURES_PATH)
model = joblib.load(MODEL_PATH)
predictions = create_customer_predictions(model, features)
predictions.to_csv(CUSTOMER_PREDICTIONS_PATH, index=False)
predictions.head()
```

## Build Risk Segments

Risk thresholds are business rules and can be adjusted after validation.

```python
segments = build_customer_segments(features, predictions)
segments.to_csv(CUSTOMER_SEGMENTS_PATH, index=False)
segments[['CustomerID', 'ChurnProbability', 'RiskSegment', 'RevenueAtRisk', 'BusinessSegment', 'RetentionRecommendation']].head(10)
```

```python
summary = create_business_summary(segments)
summary.to_csv(BUSINESS_SUMMARY_PATH, index=False)
importance = pd.read_csv(FEATURE_IMPORTANCE_PATH)
create_business_figures(segments, importance)
summary
```
